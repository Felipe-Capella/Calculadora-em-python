```python
nome = input("Digite seu nome: ").strip()
print(f"\nOlá, {nome}! Vamos iniciar a calculadora.\n")

while True:
    print("Escolha a operação:")
    print("1 - Soma (+)")
    print("2 - Subtração (-)")
    print("3 - Multiplicação (*)")
    print("4 - Divisão (/)")
    print("0 - Sair")

    opcao = input("\nDigite o número da operação: ").strip()

    if opcao == "0":
        print(f"\nEncerrando calculadora. Até mais, {nome}!")
        break

    if opcao not in ["1", "2", "3", "4"]:
        print("\nOpção inválida! Tente novamente.\n")
        continue

    try:
        num1 = float(input("Digite o primeiro número: "))
        num2 = float(input("Digite o segundo número: "))
    except ValueError:
        print("\nErro: digite apenas valores numéricos válidos.\n")
        continue

    if opcao == "1":
        resultado = num1 + num2
        print(f"\nResultado: {num1} + {num2} = {resultado}\n")
    elif opcao == "2":
        resultado = num1 - num2
        print(f"\nResultado: {num1} - {num2} = {resultado}\n")
    elif opcao == "3":
        resultado = num1 * num2
        print(f"\nResultado: {num1} * {num2} = {resultado}\n")
    elif opcao == "4":
        if num2 == 0:
            print("\nErro: não é possível dividir por zero.\n")
        else:
            resultado = num1 / num2
            print(f"\nResultado: {num1} / {num2} = {resultado}\n")
```
